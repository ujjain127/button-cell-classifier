- **Automated Sorting**: Directing batteries to appropriate processing stations
- **Defect Detection**: Identifying irregularities in battery appearance

Let's start by importing the necessary libraries.

```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, models, transforms
from torch.utils.data import DataLoader, WeightedRandomSampler
from sklearn.metrics import classification_report, confusion_matrix
from collections import Counter
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
```

## Setup Environment

We'll check if a GPU is available for training and specify the data directory containing our button cell images. PyTorch can leverage CUDA-compatible GPUs to significantly accelerate deep learning training. If no GPU is available, the code will automatically fall back to CPU processing.

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
data_dir = "button_cells"
```

## Data Preparation

Proper data preparation is crucial for training an effective deep learning model. This section covers several important preprocessing steps to ensure our dataset is clean, well-balanced, and optimized for training.

### 1. Clean Corrupt Images

First, we need to make sure our dataset doesn't contain any corrupt images that could cause issues during training. The following function walks through all image files and attempts to open them. If an image can't be opened properly, it is removed from the dataset. This step helps prevent training interruptions and ensures data quality.

```python
def clean_corrupt_images(folder_path):
    for root, _, files in os.walk(folder_path):
        for file in files:
            if file.lower().endswith(('.jpg', '.jpeg', '.png')):
                img_path = os.path.join(root, file)
                try:
                    with Image.open(img_path) as img:
                        img.verify()
                except:
                    print(f"Removing corrupt image: {img_path}")
                    os.remove(img_path)

clean_corrupt_images(data_dir)
```

### 2. Data Augmentation

To increase the effective size of our dataset and improve model generalization, we apply several data augmentation techniques:
- **Random cropping**: Resizes images to 224×224 pixels (standard ResNet input size) while maintaining important features
- **Random horizontal and vertical flips**: Creates mirror images, helping the model become invariant to orientation
- **Random rotation up to 20 degrees**: Simulates different camera angles or battery positions
- **Color jitter**: Applies variations in brightness, contrast, saturation, and hue to simulate different lighting conditions

#### Technical Implementation

The augmentations are applied randomly during training, which means:
1. Each time an image is loaded, it undergoes a different combination of transformations
2. Over many epochs, the model sees many variations of each original image
3. This effectively increases our dataset size and helps prevent overfitting

We also normalize the images using the ImageNet mean and standard deviation values ([0.485, 0.456, 0.406] and [0.229, 0.224, 0.225]), which helps with transfer learning from the pretrained ResNet50 model. Normalization standardizes the input features, leading to faster convergence during training.

```python
transform = transforms.Compose([
    transforms.RandomResizedCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.RandomRotation(20),
    transforms.ColorJitter(0.3, 0.3, 0.3, 0.1),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])
```

### 3. Load Dataset

We use PyTorch's `ImageFolder` to load our dataset. This automatically assigns class labels based on the folder structure:
- Images in `button_cells/top_view/` are assigned class 0
- Images in `button_cells/side_view/` are assigned class 1
- Images in `button_cells/bottom_view/` are assigned class 2

```python
dataset = datasets.ImageFolder(root=data_dir, transform=transform)
class_names = dataset.classes
```

### 4. Train-Validation Split

We split our dataset into training (80%) and validation (20%) sets to properly evaluate our model's performance during training. The training set is used to update the model weights, while the validation set provides an unbiased evaluation of model performance during the training process. This helps us monitor for overfitting and select the best model version.

```python
train_size = int(0.8 * len(dataset))
val_size = len(dataset) - train_size
train_data, val_data = torch.utils.data.random_split(dataset, [train_size, val_size])
```

### 5. Class Balancing

It's important to handle class imbalance to prevent the model from being biased towards the most frequent class. We use `WeightedRandomSampler` to give each class an equal chance of being selected during training, regardless of how many samples it has.

For each sample, we assign a weight that's inversely proportional to its class frequency. This means samples from underrepresented classes have a higher probability of being selected in each batch, effectively balancing the class distribution during training.

#### Example

If our dataset contains:
- 100 top view images (class 0)
- 50 side view images (class 1)
- 30 bottom view images (class 2)

The weights would be approximately:
- Top view: 1/100 = 0.01
- Side view: 1/50 = 0.02
- Bottom view: 1/30 = 0.033

This means during random sampling, bottom view images are about 3.3 times more likely to be selected than top view images, which compensates for their lower frequency in the dataset. As a result, each class has roughly equal representation in each training batch.

```python
class_counts = Counter([dataset.targets[i] for i in train_data.indices])
weights = [1.0 / class_counts[dataset.targets[i]] for i in train_data.indices]
sampler = WeightedRandomSampler(weights, len(weights))

train_loader = DataLoader(train_data, batch_size=16, sampler=sampler)
val_loader = DataLoader(val_data, batch_size=16, shuffle=False)
```

## Model Architecture

We use a ResNet50 architecture pretrained on ImageNet. ResNet (Residual Network) is a powerful CNN architecture that uses skip connections (or "shortcuts") to solve the vanishing gradient problem in deep networks. These skip connections allow the gradient to flow through the network more easily during backpropagation, enabling training of much deeper networks.

### ResNet Advantages for Battery Classification

- **Deep Feature Extraction**: ResNet50's depth allows it to learn hierarchical features from simple edges to complex battery characteristics
- **Transfer Learning Efficiency**: By leveraging ImageNet pretraining, we benefit from learned features even with limited battery data
- **Residual Blocks**: The skip connections help maintain performance even in a 50-layer network, preventing degradation
- **Batch Normalization**: Built-in normalization layers improve training stability and convergence

We replace the final fully connected layer to output 3 classes instead of the original 1000 ImageNet classes. This technique, known as transfer learning, leverages the feature extraction capabilities learned on a large dataset (ImageNet) and adapts them to our specific task with much less training data.

We set `requires_grad=True` for all parameters to fine-tune the entire network rather than just the final layer. This approach allows the model to adapt the lower-level features to our specific dataset, potentially improving performance at the cost of more training time and computational resources.

```python
model = models.resnet50(pretrained=True)
for param in model.parameters():
    param.requires_grad = True  # fine-tune all layers
model.fc = nn.Linear(model.fc.in_features, len(class_names))
model = model.to(device)
```

## Model Training

### 1. Training Setup

We use Cross-Entropy Loss, which is suitable for multi-class classification problems. It measures the difference between the predicted probability distribution and the actual class labels.

For optimization, we use Adam optimizer with a small learning rate (1e-5) for fine-tuning. Adam (Adaptive Moment Estimation) combines the benefits of two other optimizers: AdaGrad and RMSProp. It adapts the learning rate for each parameter based on both first-order (mean) and second-order (uncentered variance) moments of the gradients.

We also initialize lists to track training and validation metrics across epochs, which will help us visualize the learning progress.

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-5)

best_val_acc = 0
train_losses, val_losses = [], []
train_accuracies, val_accuracies = [], []
```

### 2. Training Loop

This is the main training loop for our model. For each epoch, we:
1. Set the model to training mode (`model.train()`), which enables gradient computation and updates to batch normalization statistics
2. Iterate through batches of training data from our DataLoader
3. Perform forward pass (compute model predictions)
4. Calculate loss between predictions and ground truth labels
5. Zero out previous gradients (`optimizer.zero_grad()`) to prevent accumulation
6. Backpropagate the loss (`loss.backward()`) to compute gradients
7. Update model parameters using the optimizer (`optimizer.step()`)
8. Calculate and store training metrics (loss and accuracy)

This process is repeated for a fixed number of epochs (20 in our case) to iteratively improve the model's performance.

```python
for epoch in range(20):
    # Training
    model.train()
    train_loss, train_correct, train_total = 0.0, 0, 0

    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        train_loss += loss.item()
        _, preds = torch.max(outputs, 1)
        train_total += labels.size(0)
        train_correct += (preds == labels).sum().item()
    epoch_loss = train_loss / len(train_loader)
    epoch_acc = 100 * train_correct / train_total
    train_acc = 100 * train_correct / train_total
    train_losses.append(train_loss / len(train_loader))
    train_accuracies.append(train_acc)
    print(f"Epoch [{epoch+1}/10] Loss: {epoch_loss:.4f} | Accuracy: {epoch_acc:.2f}%")
```

### 3. Validation

After each training epoch, we evaluate the model on the validation set:
1. Set the model to evaluation mode (`model.eval()`), which disables dropout and uses fixed batch normalization statistics
2. Disable gradient calculations using `torch.no_grad()` for memory efficiency and faster computation
3. Calculate validation loss and accuracy on unseen data
4. Save the model if it achieves better validation accuracy than previously seen (early stopping technique)

Validation is crucial to ensure that our model generalizes well to new, unseen data rather than simply memorizing the training examples. The best model is saved based on validation accuracy, not training accuracy, to avoid overfitting.

```python
model.eval()
val_loss, val_correct, val_total = 0.0, 0, 0

with torch.no_grad():
    for images, labels in val_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        loss = criterion(outputs, labels)

        val_loss += loss.item()
        _, preds = torch.max(outputs, 1)
        val_total += labels.size(0)
        val_correct += (preds == labels).sum().item()

val_acc = 100 * val_correct / val_total
val_losses.append(val_loss / len(val_loader))
val_accuracies.append(val_acc)

print(f"Epoch [{epoch+1}/20] Train Acc: {train_acc:.2f}% | Val Acc: {val_acc:.2f}%")

# Save best model
if val_acc > best_val_acc:
    best_val_acc = val_acc
    torch.save(model.state_dict(), "best_model_resnet50.pth")
```

We plot the training and validation accuracy/loss curves to visualize the model's learning progress over time. These visualizations help us identify several important patterns:

Analyzing these curves helps us make informed decisions about model hyperparameters and training duration.

### 3. Detailed Evaluation Metrics

We load the best model saved during training and perform a detailed evaluation on the validation set. We compute and visualize two key performance indicators:

1. **Classification Report**: Shows precision, recall, F1-score, and support for each class:
   - **Precision**: The proportion of positive identifications that were actually correct (TP/(TP+FP))
   - **Recall**: The proportion of actual positives that were identified correctly (TP/(TP+FN))
   - **F1-score**: The harmonic mean of precision and recall, providing a balance between the two
   - **Support**: The number of samples in each class

2. **Confusion Matrix**: A table showing the number of correct and incorrect predictions for each class. The diagonal elements represent correct predictions, while off-diagonal elements show misclassifications. This helps identify which classes are most frequently confused with each other.

```python
model.load_state_dict(torch.load("best_model_resnet50.pth"))
model.eval()

all_preds, all_labels = [], []

with torch.no_grad():
    for images, labels in val_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, preds = torch.max(outputs, 1)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

print(classification_report(all_labels, all_preds, target_names=class_names))

cm = confusion_matrix(all_labels, all_preds)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
# Show misclassified images
import numpy as np
import matplotlib.pyplot as plt

inv_transform = transforms.Compose([
    transforms.Normalize(mean=[-0.485/0.229, -0.456/0.224, -0.406/0.225],
                         std=[1/0.229, 1/0.224, 1/0.225])
])

wrong = []

model.eval()
with torch.no_grad():
    for inputs, labels in val_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        outputs = model(inputs)
        _, preds = torch.max(outputs, 1)

        for i in range(len(preds)):
            if preds[i] != labels[i]:
                wrong.append((inputs[i].cpu(), preds[i].cpu(), labels[i].cpu()))

# Show first 5 misclassified images
for i in range(min(5, len(wrong))):
    image, pred, label = wrong[i]
    image = inv_transform(image).permute(1, 2, 0).numpy().clip(0, 1)
    plt.imshow(image)
    plt.title(f"Pred: {class_names[pred]}, Actual: {class_names[label]}")
    plt.axis('off')
    plt.show()
```

### 6. Single Image Prediction

Let's create a function to make predictions on individual images and test it on a sample image. This utility function demonstrates how the model can be used for inference on new images. It handles the entire prediction pipeline:

1. Loading and converting the image to RGB
2. Applying the same transformations used during training
3. Adding a batch dimension (as models expect batches of images)
4. Running the model to get predictions
5. Converting the numerical prediction to a human-readable class name

This function can be integrated into various applications that need battery classification capabilities.

```python
from PIL import Image

def predict_image(image_path, model, transform):
    img = Image.open(image_path).convert("RGB")
    img = transform(img).unsqueeze(0).to(device)

    model.eval()
    with torch.no_grad():
        outputs = model(img)
        _, pred = torch.max(outputs, 1)
    return class_names[pred.item()]

# Example
test_image_path = "/content/drive/MyDrive/button_cells/back_view/15.jpg"
print("Prediction:", predict_image(test_image_path, model, transform))
```

## Real-time Webcam Classification

This code implements a complete real-time classification system using a webcam feed. The process flow is as follows:

1. **Capture**: Capture a frame from the webcam
2. **Preprocess**: 
   - Crop the center to get a square image (maintaining aspect ratio)
   - Convert from OpenCV's BGR format to PIL's RGB format
   - Apply the preprocessing transformations
3. **Inference**: 
   - Make a prediction using our model
   - Convert prediction index to class name
4. **Visualization**:
   - Display the result on the frame
   - Show the processed frame in a window
5. **Control**: Quit when 'q' is pressed

### Performance Considerations

For real-time applications, several factors affect performance:

- **Frame Rate**: The system typically achieves 20-30 FPS on standard hardware
- **Latency**: There is a small delay between image capture and classification (50-100ms)
- **Lighting**: Consistent, diffuse lighting improves classification accuracy
- **Camera Position**: The camera should have a clear, unobstructed view of the battery

### Deployment Considerations

When deploying the real-time classification system, consider the following:

- **Hardware Requirements**: Ensure the hardware meets the necessary specifications for real-time processing
- **Software Dependencies**: All required libraries and dependencies must be installed on the deployment machine
- **Model Optimization**: The classification model may need optimization for faster inference on different hardware
- **Integration**: The system should be properly integrated with the hardware components (e.g., camera, lighting)

This implementation demonstrates how the trained model can be integrated into a practical application for real-time button cell classification. The same approach can be adapted for industrial settings by replacing the webcam with industrial cameras and adding appropriate hardware integration.

```python
import cv2
from PIL import Image
import torch
from torchvision import transforms

# Preprocessing
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3)
])

# Load model
model.eval()  # your model must be defined and loaded
class_names = ['top_view', 'side_view', 'bottom_view','background']

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Crop center square
    h, w, _ = frame.shape
    min_dim = min(h, w)
    center_crop = frame[(h - min_dim)//2:(h + min_dim)//2,
                        (w - min_dim)//2:(w + min_dim)//2]

    # Convert to PIL and preprocess
    img = Image.fromarray(cv2.cvtColor(center_crop, cv2.COLOR_BGR2RGB))
    input_tensor = transform(img).unsqueeze(0)  # Add batch dim

    # Predict
    with torch.no_grad():
        output = model(input_tensor)
        _, pred = torch.max(output, 1)
        label = class_names[pred.item()]

    # Display
    cv2.putText(frame, f"Prediction: {label}", (30, 50),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
    cv2.imshow("Button Cell Classifier", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader
from sklearn.metrics import classification_report
from PIL import Image
import cv2
from collections import Counter
import numpy as np

# Config
DATA_DIR = "./button_cells"
BATCH_SIZE = 32
EPOCHS = 5
NUM_CLASSES = 3  # Set to 2 if doing binary classification
IMG_SIZE = 224
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Transforms
train_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomRotation(30),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.3, contrast=0.3),
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3)
])

val_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3)
])

# Datasets & Loaders
dataset = datasets.ImageFolder(root=DATA_DIR, transform=train_transform)
class_names = dataset.classes
train_size = int(0.8 * len(dataset))
val_size = len(dataset) - train_size
train_dataset, val_dataset = torch.utils.data.random_split(dataset, [train_size, val_size])
val_dataset.dataset.transform = val_transform

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE)

# Model
model = models.resnet50(pretrained=True)
for param in model.parameters():
    param.requires_grad = False
model.fc = nn.Sequential(
    nn.Linear(model.fc.in_features, 128),
    nn.ReLU(),
    nn.Dropout(0.3),
    nn.Linear(128, NUM_CLASSES)
)
model = model.to(DEVICE)

# Loss & Optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.fc.parameters(), lr=0.001)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max', factor=0.3, patience=2)

# Training Loop
for epoch in range(EPOCHS):
    model.train()
    running_loss = 0
    correct = 0

    for inputs, labels in train_loader:
        inputs, labels = inputs.to(DEVICE), labels.to(DEVICE)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        correct += (outputs.argmax(1) == labels).sum().item()

    acc = correct / len(train_loader.dataset)

    # Validation
    model.eval()
    val_correct = 0
    y_true, y_pred = [], []
    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs, labels = inputs.to(DEVICE), labels.to(DEVICE)
            outputs = model(inputs)
            preds = outputs.argmax(1)
            val_correct += (preds == labels).sum().item()
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(preds.cpu().numpy())

    val_acc = val_correct / len(val_loader.dataset)
    scheduler.step(val_acc)
    print(f"Epoch {epoch+1}/{EPOCHS}, Train Acc: {acc:.2f}, Val Acc: {val_acc:.2f}")

# Save Model
torch.save(model.state_dict(), "side_view_model.pth")
print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=class_names))

# Live Webcam Prediction (Jupyter or script)
model.eval()
model.load_state_dict(torch.load("side_view_model.pth", map_location=DEVICE))

transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3)
])
```

```python
seen_views = set()
required_views = set(class_names)  # {'top_view', 'bottom_view', 'side_view'}
frame_counter = 0
max_frames = 300  # ~10 seconds if running at ~30 FPS
```

```python
cap = cv2.VideoCapture(0)
print("Press 'q' to quit")

while True:
    ret, frame = cap.read()
    if not ret:
        print("❌ Failed to capture frame")
        break

    h, w, _ = frame.shape
    min_dim = min(h, w)
    crop = frame[(h - min_dim)//2:(h + min_dim)//2, (w - min_dim)//2:(w + min_dim)//2]
    img = Image.fromarray(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
    input_tensor = transform(img).unsqueeze(0).to(DEVICE)

    with torch.no_grad():
        output = model(input_tensor)
        pred = output.argmax(1)
        probs = torch.softmax(output, dim=1).squeeze().cpu().numpy()

    label = class_names[pred.item()]
    conf = probs[pred.item()] * 100
    seen_views.add(label)
    frame_counter += 1

    # Show current view
    cv2.putText(frame, f"{label} ({conf:.1f}%)", (30, 50),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

    # Show which views are seen
    views_msg = "Seen: " + ", ".join(sorted(seen_views))
    cv2.putText(frame, views_msg, (30, 90),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)

    # All views seen
    if seen_views == required_views:
        cv2.putText(frame, "✅ All views detected!", (30, 130),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)
        seen_views.clear()
        frame_counter = 0

    # Timeout warning
    elif frame_counter > max_frames:
        missing = required_views - seen_views
        if missing:
            missing_msg = "Missing: " + ", ".join(sorted(missing))
            cv2.putText(frame, missing_msg, (30, 130),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 0, 255), 2)

    cv2.imshow("Button Cell Classification", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
